```python
# 1. Online Shopping Discount
# Ask the user to enter their shopping amount.
# If the amount is 2000 or more, print "Discount Applied".
# Otherwise, print "No Discount".

shopping = int(input("enter amount: "))
if shopping >= 2000:
    print("discount applied")
else:
    print("no discount")
```

```python
# 2. Username and Password
# Ask the user to enter a username and password.
# If username is "admin" AND password is "1234",
# print "Login Successful".
# Otherwise, print "Invalid Credentials".

username = input("enter your username: ")
password = input("enter your password: ")
if username =="admin" and password == "1234":
    print("login sucessful")
else:
    print("invalid credentials")
```

```python
# 3. Movie Ticket Eligibility
# Ask the user for their age.
# If age is below 13, print "Child Ticket".
# If age is between 13 and 17, print "Teen Ticket".
# Otherwise, print "Adult Ticket".

age = int(input("enter your age: "))
if age < 13:
    print("child ticket")
elif age >= 13 and age <= 17:
    print("teen ticket")
else:
    print("adult ticket")
```

```python
# 4. Weather Activity
# Ask the user to enter the weather:
# "sunny", "rainy", or "cloudy".
# If sunny, print "Go for a walk".
# If rainy, print "Stay indoors".
# If cloudy, print "You can go outside".
# For any other input, print "Unknown Weather".

weather = input("enter the weather: ")
if weather == "sunny":
    print("go for a walk")
elif weather == "rainy":
    print("stay indoor")
elif weather == "cloudy":
    print("you can go outside")
else:
    print("unknown weather")
```

```python
# 5. College Attendance
# Ask the user to enter their attendance percentage.
# If attendance is below 40, print "Meet the HOD".
# If attendance is 40 or above but below 75,
# print "Not Eligible for Exam".
# If attendance is 75 or above,
# print "Eligible for Exam".

attendance = int(input("enter your attendance percentag: "))
if attendance < 40:
    print("meet hod")
elif attendance < 75:
    print("not eligible for exam")
else:
    print("eligible for exam")
```

```python
# 6. Game Access
# Ask the user whether they have an account
# and whether they have completed registration.
# If they have an account AND completed registration,
# print "You can play".
# Otherwise, print "Complete your registration".

account = input("do you have a game account? (yes/no)")
registration = input("have you completed registration? (yes/no)")
if account == "yes" and registration == "yes":
    print("you can play")
else:
    print("complete your registration")
```

```python
# 7. Delivery Charges
# Ask the user to enter their order amount.
# If the amount is 1000 or more,
# print "Free Delivery".
# If the amount is between 500 and 999,
# print "Delivery Charge: 40".
# Otherwise, print "Delivery Charge: 80".

amount = int(input("enter your order amount: "))
if amount >= 1000:
    print("free delivery")
elif amount >= 500 and amount < 1000:
    print("delivery charge: 40")
else:
    print("delivery charge: 80")
```

```python
# 8. Friendship Invitation
# Ask the user whether they are free today
# and whether they want to go out.
# If they are free AND want to go out,
# print "Let's go out!".
# If they are free but do not want to go out,
# print "Maybe another day".
# Otherwise, print "Let's plan another time".

invitation = input("do you have free time? (yes/no)")
out = input("do you want to go out? (yes/no)")
if invitation == "yes" and out == "yes":
    print("let's go out")
elif invitation == "yes" and out == "no":
    print("maybe another day")
else:
    print("let's plan another day")
```

```python
# 9. Password Strength Checker
# Ask the user to enter a password.
# If its length is less than 6, print "Weak Password".
# If its length is between 6 and 10,
# print "Medium Password".
# If its length is greater than 10,
# print "Strong Password".
# Hint: Use len(password).

password = input("enter your password")
if len(password) < 6:
    print(f"{password} is a weak password")
elif len(password) <= 10:
    print(f"{password} is a medium password")
else:
    print(f"{password} is a strong password")
```

```python
# 10. College Club Selection
# Ask the user which club they want to join:
# "coding", "sports", or "music".
# If coding, ask whether they know Python.
# If they know Python, print "Welcome to Coding Club".
# Otherwise, print "Learn Python first".
# If sports, print "Welcome to Sports Club".
# If music, print "Welcome to Music Club".
# For any other choice, print "Club not available".
# Hint: Use nested if-else statements.

club = input("which club do you want to join? (coding/sports/music)")
if club == "coding":
    python = input("do you know python?(yes/no)")
    if python == "yes":
        print("welcome to coding club")
    else:
        print("learn python first")
elif club == "music":
    print("welcome to music club")
elif club == "sports":
    print("welcome to sports club")
else:
    print("club not found")
```
