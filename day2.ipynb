```python
# Q.1 Choose a fruit
# Ask the user to enter a fruit name.
# If it is "apple", print "Red fruit".
# If it is "banana", print "Yellow fruit".
# Otherwise, print "Fruit not recognised".

fruit = input("enter a fruit: ")
if fruit == "apple":
    print("red fruit")
elif fruit == "banana":
    print("yellow fruit")
else:
    print("fruit not recognised")
```

```python
# Q2. Favourite subject
# Ask the user to enter their favourite subject.
# If it is "python", print "Great choice!".
# If it is "maths", print "Interesting!".
# Otherwise, print "That's nice too!".

subject = input("enter your fav sub: ")
if subject == "python":
    print("great choice")
elif subject == "maths":
    print("interesting")
else:
    print("that is nice too")
```

```python
# Q3. Choose a drink
# Ask the user to enter a drink.
# If it is "tea", print "Tea time!".
# If it is "coffee", print "Coffee time!".
# Otherwise, print "Enjoy your drink!".

drink = input("enter a drink: ")
if drink == "tea":
    print(f"it is {drink} time")
elif drink == "coffee":
    print(f"it is {drink} time")
else:
    print(f"enjoy your {drink}")
```

```python
# Q4. Check the animal
# Ask the user to enter an animal name.
# If it is "dog", print "Woof woof!".
# If it is "cat", print "Meow!".
# Otherwise, print "That's an interesting animal!".

animal = input("enter an animal name: ")
if animal == "dog":
    print("woof woof")
elif animal == "cat":
    print("meow")
else:
    print(f"that is an interesting choise i like {animal}")
```

```python
# Q5. Greeting program
# Ask the user to enter a greeting.
# If it is "hello", print "Hi there!".
# If it is "good morning", print "Good morning to you too!".
# Otherwise, print "Nice to meet you!".

greeting = input("enter your greeting: ")
name = input("enter a name: ")
if greeting == "hello":
    print(f"hello there {name}")
elif greeting == "good morning":
    print(f"good mornig to you too {name}")
else:
    print(f"nice to meet you {name}")
```
