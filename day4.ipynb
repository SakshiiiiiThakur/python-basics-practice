```python
#  - QUESTION 1 ---> marks percentage calculator

maths=int(input("enter your marks: "))
english=int(input("enter your marks: "))
hindi=int(input("enter your marks: "))

total=maths+english+hindi
avg=total/3
print(avg)

if avg>30 and avg<=60:
    print("pass by c grade")
elif avg>60 and avg<=80:
    print("pass by b grade")
elif avg>80 and avg<=100:
    print("pass by a grade")
else:
    print("fail")
```

```python
# QUESTION 4: Sum of Natural Numbers
# Take a number n as input.
# Calculate the sum of numbers from 1 to n using a for loop.
# Example: n = 5
# Output: 15

num=int(input("enter your number: "))
sum=0
for i in range(1,num+1):
    sum=sum+i
print(sum)
```

```python
# QUESTION 5: Multiplication Table
# Take a number as input.
# Print its multiplication table from 1 to 10.
# Example: n = 5
# Output:
# 5 x 1 = 5
# 5 x 2 = 10
# Continue up to 5 x 10 = 50.

num=int(input("enter your number: "))
for i in range(1,11):
    print(num,"x",i,"=",num*i)
```

```python
# QUESTION 6: Count Numbers Divisible by 3
# Take a number n as input.
# Count how many numbers from 1 to n are divisible by 3.
# Example: n = 10
# Output: 3

num=int(input("enter your number: "))
count=0
for i in range(1,num+1):
    if i%3==0:
        count=count+1
        print(count)
```
